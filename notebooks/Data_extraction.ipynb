```python
max_unique_values = 20
```

```python
# Diccionario con columnas incluidas (categóricas o con pocos valores únicos)
included = {}
# Lista de columnas excluidas con número de valores únicos
excluded = []
```

```python
for col in adata.obs.columns:
    nunique = adata.obs[col].nunique()
    dtype = adata.obs[col].dtype.name

    if dtype in ["category", "object"] or nunique <= max_unique_values:
        included[col] = adata.obs[col].unique().tolist()
    else:
        excluded.append({
            "column": col,
            "unique_values": nunique
        })
```

```python
with open("categorical_unique_values.json", "w") as f:
    json.dump(included, f, indent=4)
```

```python
with open("excluded_columns.json", "w") as f:
    json.dump(excluded, f, indent=4)
```
